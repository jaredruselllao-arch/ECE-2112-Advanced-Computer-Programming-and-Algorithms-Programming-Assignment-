A. WORD ROTATION PROBLEM


```python
def rotate_word(word): #This line of code defines the function and names it rotate_word, alongside creating parameters 
    return (word[1:] + word[0]) #This is the return function that gives back the 1st index or the 2nd element of the provided word, and putting the 0th index at the end of the string, while being combined with the concept of concatenation which is just adding string together which is what allowed for the space between to be removed
```

```python
print(rotate_word("Python")) #These are the samples to show the results
print(rotate_word("Logic")) #These are the samples to show the results
print(rotate_word("Code")) #These are the samples to show the results
print(rotate_word("A")) #These are the samples to show the results
```

USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name): #Creates and Defines the function make)username with 2 paramets
    return (first_name.lower().replace(" ","")+"."+last_name.lower().replace(" ","")) #Returns the first value given, as all lowercases, while using concatenation to add the middle period while removing the space, then follows by adding the second given element, while being all lowewrcased 
```

```python
print(make_username("Ada", "Lovelace")) #These are the samples to show the results
print(make_username("Alan", "Turing")) #These are the samples to show the results
print(make_username("Ana Maria", "De Leon")) #These are the samples to show the results
```

C. BOOKEND SWAP PROBLEM


```python
def swap_bookends(items): #Defines the funciton swap_bookends with its corresponding parameters
    first, *middle, last = items #The line of code that allows us to use 3 arguments; instead of only one by unpacking the 3 terms
    return ([last] + middle + [first]) #Returns the Last given value, then everything in the middle, then the first value
```

```python
print(swap_bookends([1,2,3,4,5,6])) #These are the samples to show the results
print(swap_bookends(["red", "green", "blue"])) #These are the samples to show the results
print(swap_bookends([8, 3])) #These are the samples to show the results
```
